```python
# ==========================================
# MINOR PROJECT 5
# CARD READER SIMULATOR
# ==========================================

print("===================================")
print("       CARD READER SIMULATOR")
print("===================================")

# Taking dummy card details
card_number = input("Enter dummy card number: ").replace(" ", "")
card_holder = input("Enter card holder name: ")
expiry = input("Enter expiry date (MM/YY): ")
cvv = input("Enter dummy CVV: ")

# Check card type
if card_number.startswith("4"):
    card_type = "Visa"

elif card_number.startswith("5"):
    card_type = "Mastercard"

elif card_number.startswith("3"):
    card_type = "American Express"

else:
    card_type = "Unknown"

# Basic validation
number_valid = card_number.isdigit() and len(card_number) in [13, 15, 16, 19]
cvv_valid = cvv.isdigit() and len(cvv) in [3, 4]
expiry_valid = len(expiry) == 5 and expiry[2] == "/"

print("\n===================================")
print("           CARD DETAILS")
print("===================================")

print("Card Holder :", card_holder)
print("Card Type   :", card_type)

if number_valid:
    print("Card Number : Valid format")
else:
    print("Card Number : Invalid format")

if expiry_valid:
    print("Expiry Date : Valid format")
else:
    print("Expiry Date : Invalid format")

if cvv_valid:
    print("CVV         : Valid format")
else:
    print("CVV         : Invalid format")

# Final result
if number_valid and cvv_valid and expiry_valid:
    print("\nCard Status : ACCEPTED")
    print("Transaction : Ready for processing")

else:
    print("\nCard Status : REJECTED")
    print("Transaction : Invalid card details")

print("===================================")
```
