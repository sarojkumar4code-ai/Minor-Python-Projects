```python
import requests

print("====================================")
print("       RANDOM CAT FACT GENERATOR")
print("====================================")

url = "https://catfact.ninja/fact"

response = requests.get(url)

if response.status_code == 200:

    data = response.json()

    fact = data["fact"]

    print("\n🐱 Cat Fact:")
    print(fact)

else:

    print("\nUnable to fetch cat fact.")

print("\n====================================")
```
