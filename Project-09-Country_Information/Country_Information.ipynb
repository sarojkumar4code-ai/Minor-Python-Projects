```python
import requests

print("====================================")
print("       COUNTRY INFORMATION FINDER")
print("====================================")

country = input("Enter country name: ")

url = f"https://restcountries.com/v3.1/name/{country}"

response = requests.get(url)

if response.status_code == 200:

    data = response.json()
    country_data = data[0]

    name = country_data["name"]["common"]
    capital = country_data.get("capital", ["Not available"])[0]
    region = country_data.get("region", "Not available")
    population = country_data.get("population", "Not available")

    # Get currency
    currencies = country_data.get("currencies", {})
    currency_name = list(currencies.values())[0]["name"] if currencies else "Not available"

    # Get languages
    languages = country_data.get("languages", {})
    language_list = ", ".join(languages.values()) if languages else "Not available"

    print("\n====================================")
    print("       COUNTRY INFORMATION")
    print("====================================")

    print("Country     :", name)
    print("Capital     :", capital)
    print("Region      :", region)
    print("Population  :", population)
    print("Currency    :", currency_name)
    print("Languages   :", language_list)

else:
    print("\nCountry not found or API error.")

print("====================================")
```
