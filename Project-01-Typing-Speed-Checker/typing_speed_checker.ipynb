```python
import time

print("========first project======")
print("======TYPING SPEED CHECKER======")

text = "Python is a simple and powerful programming language."

print("\nType the following sentence:")
print("\n", text)

input("\nPress ENTER when you are ready...")

print("\nStart typing:")

start_time = time.time()

user_text = input()

end_time = time.time()

# Calculate time
time_taken = end_time - start_time

# Calculate words
words = len(user_text.split())

# Calculate typing speed
if time_taken > 0:
    wpm = (words / time_taken) * 60
else:
    wpm = 0

# Calculate correct characters
correct = 0

for i in range(min(len(text), len(user_text))):
    if text[i] == user_text[i]:
        correct += 1

# Accuracy
if len(text) > 0:
    accuracy = (correct / len(text)) * 100
else:
    accuracy = 0

# Incorrect characters
incorrect = len(text) - correct


print("RESULT")

print("Time Taken      :", round(time_taken, 2), "seconds")
print("Typing Speed    :", round(wpm, 2), "WPM")
print("Accuracy        :", round(accuracy, 2), "%")
print("Correct Chars   :", correct)
print("Incorrect Chars :", incorrect)

```
