```python
import requests

print("====================================")
print("       CURRENCY CONVERTER")
print("====================================")

amount = float(input("Enter amount: "))

from_currency = input("Enter source currency (e.g. INR): ").upper()
to_currency = input("Enter target currency (e.g. USD): ").upper()

url = f"https://api.frankfurter.dev/v2/rate/{from_currency}/{to_currency}"

response = requests.get(url)

if response.status_code == 200:

    data = response.json()

    rate = data["rate"]
    converted_amount = amount * rate

    print("\n====================================")
    print("          CONVERSION RESULT")
    print("====================================")

    print("From Currency :", from_currency)
    print("To Currency   :", to_currency)
    print("Exchange Rate :", rate)
    print("Amount        :", amount)
    print("Converted     :", round(converted_amount, 2), to_currency)

else:
    print("\nInvalid currency code or API error.")

print("====================================")
```
