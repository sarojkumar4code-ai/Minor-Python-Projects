```python
import requests

print("====================================")
print("        RANDOM JOKE GENERATOR")
print("====================================")

url = "https://official-joke-api.appspot.com/random_joke"

response = requests.get(url)

if response.status_code == 200:

    data = response.json()

    setup = data["setup"]
    punchline = data["punchline"]

    print("\nJoke:")
    print(setup)

    input("\nPress Enter to see the punchline...")

    print("\nPunchline:")
    print(punchline)

else:
    print("Unable to fetch joke.")

print("\n====================================")
```
